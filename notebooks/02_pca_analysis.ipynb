2.2 Dimensionality Reduction - PCA (Principal Component Analysis) 

```python
from sklearn.decomposition import PCA
# Use the cleaned + encoded dataset
X_final = X_encoded.loc[df_clean.index]   # ensures same rows as df_clean
y_final = df_clean["target"]

# Apply PCA
pca = PCA()
X_pca = pca.fit_transform(X_final)

# Explained variance
explained_variance = pca.explained_variance_ratio_

# 1️⃣ Variance retained plot
plt.figure(figsize=(8,5))
plt.bar(range(1, len(explained_variance)+1), explained_variance, alpha=0.7, align="center")
plt.step(range(1, len(explained_variance)+1), np.cumsum(explained_variance), where="mid", color="red", label="Cumulative Variance")
plt.xlabel("Principal Component")
plt.ylabel("Variance Explained")
plt.title("PCA - Explained Variance Ratio")
plt.legend()
plt.show()

# 2️⃣ Scatter plot of first 2 principal components
plt.figure(figsize=(8,6))
plt.scatter(X_pca[:,0], X_pca[:,1], c=y_final, cmap="bwr", alpha=0.7)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA - First 2 Principal Components")
plt.colorbar(label="Target (0=No Disease, 1=Disease)")
plt.show()
```
